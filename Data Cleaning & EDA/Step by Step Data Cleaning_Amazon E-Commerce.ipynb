# 4. Data Preparation

## 4.1 Column & Data Type

```python
#memeriksa nama-nama kolom
print(df.columns.tolist())
#Interpetasi: terdapat whitespace pada nama kolom Sales Channel
#Interpretasi: nama kolom tidak konsisten (kapital dan kecil) sehingga perlu diatasi
```

```python
#mengatasi nama kolom yang tidak konsisten dan mengandung whitespace
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','-')
df = df.rename(columns={
    'sales-channel-': 'sales-channel',
    'sku': 'SKU',
    'asin': 'ASIN',
    'b2b': 'B2B',
    'unnamed:-22':'unnamed:22'
})
```

```python
#menampilkan 5 data teratas
df.head()
#Interpretasi: Kolom currency, ship-country, fulfilled-by, dan unnamed:22 memiliki nilai yang konstan sehingga tidak memiliki arti analisis
#Interpretasi: Kolom index memiliki nilai yang unik sehingga tidak memiliki arti analisis
```

```python
#menghapus kolom konstan dan kolom yang tidak memiliki arti analisis
df.drop(['index','currency','ship-country','fulfilled-by','unnamed:22'],axis=1,inplace=True)
```

```python
#memeriksa tipe data dari setiap kolom
df.dtypes
#Interpretasi: Tipe data date adalah string, sehingga perlu diubah menjadi tipe data datetime agar dapat dilakukan analisis berbasis waktu
#Interpretasi: Tipe data ship-postal-code adalah float, sehingga perlu diubah menjadi tipe data string/object karena berfungsi sebagai kode atau identitas
```

```python
#menangani kolom date dengan mengubah menjadi tipe data datetime
df['date'] = pd.to_datetime(
    df['date'],
    format='%m-%d-%y',
    errors='coerce'
)
#menangani kolom ship-postal-code dengan mengubah menjadi tipe data string/object
df['ship-postal-code']=df['ship-postal-code'].astype(str)
```

## 4.3 Data Cleaning

### a. Mengatasi Nilai Tidak Konsisten

```python
#mengatasi data tidak konsisten
df['ship-state']=df['ship-state'].str.upper()
df['ship-state'] = df['ship-state'].replace({
    'ORISSA': 'ODISHA',
    'RJ': 'RAJASTHAN',
    'RAJSHTHAN': 'RAJASTHAN',
    'RAJSTHAN': 'RAJASTHAN',
    'NL': 'NAGALAND',
    'PUNJAB/MOHALI/ZIRAKPUR': 'PUNJAB',
    'PB': 'PUNJAB',
    'APO': 'ANDHRA PRADESH',
    'AR': 'ARUNACHAL PRADESH',
    'PONDICHERRY': 'PUDUCHERRY',
    'NEW DELHI': 'DELHI'
})
#menghapus kolom ship-city, karena banyak nilai yang salah pada kolom tersebut
df.drop(['ship-city'],axis=1,inplace=True)
#menghapus kolom promotion-ids, karena banyak nilai yang salah dan merupakan kolom yang tidak memiliki arti analisis
df.drop(['promotion-ids'],axis=1,inplace=True)
```

### b. Mengatasi Duplicate Data

```python
#mengatasi data duplikat dengan menghapus salah satu baris duplikat
df.drop_duplicates(inplace=True)
```

### c. Mengatasi Missing Value

```python
#menghapus baris yang berisi missing value <5% (ship-city)
df=df.dropna(subset=['ship-state','ship-postal-code'])
#mengatasi missing value dengan mengisi menggunakan modus pada kolom Courier Status
df['courier-status']=df['courier-status'].fillna(df['courier-status'].mode()[0])
#mengatasi missing value dengan mengisi menggunakan median pada kolom Amount
df['amount']=df['amount'].fillna(df['amount'].median())
```
